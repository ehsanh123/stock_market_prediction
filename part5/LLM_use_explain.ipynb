```python
def answer_question(qu,df1):   
    # Ensure the DataFrame is not empty
    if df1.empty:
        return "The DataFrame is empty. Please upload a valid file."

    # Create the prompt 
    # This format works for mistral
    num_rows = df1.shape[0]
    Prompt = ''
    for i in range(num_rows):
        Prompt = f"""[INST] You are a data analyst. Below is a preview of the dataset:

{df1.head(num_rows-i).to_string()}

Based on this data, answer the following question: {qu} [/INST]"""
        # Tokenize the prompt
        tokens = llm.tokenize(Prompt.encode('utf-8'))
        num_tokens = len(tokens)
        if num_tokens < 2048:
            break
    
    # Generate response
    res = llm(
    prompt=Prompt,
    max_tokens=512,
    temperature=0.2,
    #0.1 to 0.3 for fouced answers
    #0.7 to 0.9 for creative answers
    stop=["</s>", "[INST]", "\n\nUser:"]
    )

    resp = res["choices"][0]["text"].strip()
    return resp
```
