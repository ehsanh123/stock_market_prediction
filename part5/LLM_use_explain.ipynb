```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler, MinMaxScaler
import joblib
import os

class FinancialDataPreprocessor:
    #set up the preprocessor with necessary scalers and column definitions
    def __init__(self):
        # Define scalers and column categories
        self.scaler_standard = StandardScaler()
        self.Standard_Column_used = []
        self.scaler_minmax_pct = MinMaxScaler()
        self.minmax_Column_used = []
        self.scaler_minmax_log = MinMaxScaler()
        self.minmax_log_Column_used = []
        

        self.Double_Columns = [
            'Revenue', 'Operating Profit', 'Profit After Tax', 'Basic ESP',
            'Interim Dividend', 'Net From Oprating', 'CapEx'
        ]
        self.PCT_Columns = [
            'Revenue', 'Operating Profit', 'Profit After Tax', 
            'Free Cash Flow', 'Avg Close Price', 'Net From Oprating', 'CapEx'
        ]
        self.Drivit_Columns = ['Operating Margin (%)', 'Profit Margin (%)', 'Debt-to-Equity(%)']

        self.Standard_Columns = [
            'Operating Profit', 'Profit After Tax', 'Basic ESP', 'Net Cash', 
            'Interim Dividend', 'Free Cash Flow', 'Operating Margin (%)_Dri', 
            'Profit Margin (%)_Dri', 'Debt-to-Equity(%)_Dri'
        ]
        self.log_Columns = ['Revenue', 'Total Equity']
        self.PCT_Cols_Final = [
            'Revenue_PCT', 'Operating Profit_PCT', 'Profit After Tax_PCT', 
            'Free Cash Flow_PCT', 'Net From Oprating_PCT', 'CapEx_PCT'
        ]

        self.final_columns = [
            'Revenue', 'Operating Profit', 'Basic ESP', 'Net Cash', 'Free Cash Flow',
            'Revenue_PCT', 'Operating Profit_PCT', 'Free Cash Flow_PCT', 
            'Net From Oprating_PCT', 'CapEx_PCT', 'Operating Margin (%)_Dri',
            'Profit Margin (%)_Dri', 'Debt-to-Equity(%)_Dri'
        ]

    # Preprocess the data by handling missing values, scaling, and creating new features
    def preprocess(self, df: pd.DataFrame) -> pd.DataFrame:
        df = df.copy()

        # 1. Double half-year values
        for idx, row in df.iterrows():
            if row['Year'] % 1 != 0:
                for col in self.Double_Columns:
                    df.at[idx, col] = row[col] * 2

        # 2. Create percentage change columns
        for col in self.PCT_Columns:
            df[col + '_PCT'] = df[col].pct_change() * 100
            df[col + '_PCT'] = df[col + '_PCT'].clip(-150, 150)
            df.loc[df['Year'] == 2015, col + '_PCT'] = 0

        # 3. Create derivative columns
        for col in self.Drivit_Columns:
            df[col + '_Dri'] = df[col].diff()

        # 4. Select relevant features
        df = df[self.final_columns + ['Avg Close Price_PCT']]  # Keep label for now

        return df

    def fit_transform(self, df: pd.DataFrame) -> pd.DataFrame:
        df = df.copy()

        for col in self.Standard_Columns:
            if col in df.columns:
                self.scaler_standard.fit(df[[col]])
                df[col] = self.scaler_standard.transform(df[[col]])
                df[col] = df[col].clip(-3, 3) / 3

        for col in self.log_Columns:
            if col in df.columns:
                df[col] = np.log1p(df[col])
                self.scaler_minmax_log.fit(df[[col]])
                df[col] = self.scaler_minmax_log.transform(df[[col]])

        for col in self.PCT_Cols_Final:
            if col in df.columns:
                self.scaler_minmax_pct.fit(df[[col]])
                df[col] = self.scaler_minmax_pct.transform(df[[col]])


        return df

    def transform(self, df: pd.DataFrame) -> pd.DataFrame:
        df = df.copy()
        
        for col in self.Standard_Columns:
            if col in df.columns:
                for col1 in self.Standard_Column_used:
                    if col1 is not col: continue
                    df[col] = self.scaler_standard.transform(df[[col]])
                    df[col] = df[col].clip(-3, 3) / 3
                    break
                # df[col] = self.scaler_standard.transform(df[[col]])
                # df[col] = df[col].clip(-3, 3) / 3

        for col in self.log_Columns:
            if col in df.columns:
                for col1 in self.minmax_log_Column_used:
                    if col1 is not col: continue

                    df[col] = np.log1p(df[col])
                    df[col] = self.scaler_minmax_log.transform(df[[col]])
                    break

        for col in self.PCT_Cols_Final:
            if col in df.columns:
                for col1 in self.minmax_Column_used:
                    if col1 is not col: continue

                    df[col] = self.scaler_minmax_pct.transform(df[[col]])
                    break

        return df
    
    def drop_nulls(self, X: pd.DataFrame, y: pd.Series):
        X = X.dropna()
        y = y.loc[X.index]
        return X, y
    
    # Prepare the target variable for classification
    def prepare_target(self, df: pd.DataFrame) -> pd.Series:
        return df['Avg Close Price_PCT'].apply(lambda pct: pct >= 0)

    # Get the final feature set
    def get_features(self):
        return self.final_columns
    
    #save and load methods for the scalers
    def save(self, path: str):
        os.makedirs(path, exist_ok=True)
        joblib.dump({
            'scaler': scaler_standard,
            'columns': self.Standard_Columns
            }, os.path.join(path, 'scaler_standard.pkl'))

        joblib.dump({
            'scaler': scaler_minmax_pct,
            'columns': self.PCT_Columns
        }, os.path.join(path, 'scaler_minmax_pct.pkl'))

        joblib.dump({
            'scaler': scaler_minmax_log,
            'columns': self.log_Columns
        }, os.path.join(path, 'scaler_minmax_log.pkl'))

    def load(self, path: str):
        # self.
        scaler_standard1 = joblib.load(os.path.join(path, 'scaler_standard.pkl'))
        self.scaler_standard = scaler_standard1['scaler']
        self.Standard_Column_used = scaler_standard1['columns']
        # self.
        scaler_minmax_pct1 = joblib.load(os.path.join(path, 'scaler_minmax_pct.pkl'))
        self.scaler_minmax_pct = scaler_minmax_pct1['scaler']
        self.minmax_Column_used = scaler_minmax_pct1['columns']
        # self.
        scaler_minmax_log1 = joblib.load(os.path.join(path, 'scaler_minmax_log.pkl'))
        self.scaler_minmax_log = scaler_minmax_log1['scaler']
        self.minmax_log_Column_used = scaler_minmax_log1['columns']
```

```python
#Load and process data from Excel files
#1- load data from excel files
import pandas as pd

path = "https://github.com/ehsanh123/stock_market_prediction/raw/main/Part4/All%20Ficnal%20data.xlsx"
excel_file = pd.ExcelFile(path)
Comapnies = excel_file.sheet_names

print(f"Found sheets: {Comapnies}")

dataframes_list = []  # For easier concatenation later
for company in Comapnies:
    df = pd.read_excel(excel_file, sheet_name=company)
    df['Company'] = company
    dataframes_list.append(df)

##############################################    
# 2- Combine all dataframes using the list
Data = pd.concat(dataframes_list, ignore_index=True)
Data1 = Data.copy()
```
